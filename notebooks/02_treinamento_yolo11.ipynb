```python
from ultralytics import YOLO
```

```python
model = YOLO('yolo11n-seg.pt')
```

```python
model.train(data='/content/drive/MyDrive/TCC/Modelo/Dataset/19_06 - Telhados+ 2048/data.yaml',
            epochs=200,
            patience=8,
            batch= 1,
            single_cls = True,
            project='/content/drive/MyDrive/TCC/Modelo/Result 1024',
            imgsz=2048)
```

```python
results = model.val(imgsz =  640,
                    save_hybrid = False,
                    split='test',
                    project= '/content/drive/MyDrive/TCC/MVP YoloV8/Val Yolov8 Dataset Augmentation')
```

```python
results_img = model.predict(source='/content/drive/MyDrive/TCC/MVP YoloV8/Telhados Dataset Yolo/test/images/506_crop_0_2_jpg.rf.22b1ed22dae94ec62e58bf175a445987.jpg' ,
                            conf = 0.25,
                            save = True,
                            save_txt=True,
                            save_conf=True,
                            save_crop=True,
                            project= 'runs/detect')
```
