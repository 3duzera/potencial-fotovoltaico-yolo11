```python
from PIL import Image
import os

def split_image(input_path, output_folder, grid_size=(2, 2)):
    """
    Divide uma imagem .tif em partes menores e salva como .jpg.
    :param input_path: Caminho da imagem .tif
    :param output_folder: Pasta para salvar os arquivos .jpg
    :param grid_size: Número de divisões (linhas, colunas)
    """
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)

    # Abrir a imagem
    image = Image.open(input_path)
    width, height = image.size

    # Calcular tamanho dos recortes
    row_count, col_count = grid_size
    tile_width = width // col_count
    tile_height = height // row_count

    # Cortar e salvar as imagens menores
    for row in range(row_count):
        for col in range(col_count):
            left = col * tile_width
            upper = row * tile_height
            right = left + tile_width
            lower = upper + tile_height

            cropped_image = image.crop((left, upper, right, lower))
            output_path = os.path.join(output_folder, f'532_crop_{row}_{col}.jpg')
            cropped_image.save(output_path, 'JPEG')
            print(f'Salvo: {output_path}')

# Exemplo de uso
split_image('532.tif', '532_dividida', grid_size=(10, 10))
```
